```python
#import libraries i need
import neuscitk as ntk
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
```

```python
#import all data
ramp_data = ntk.LabChartDataset(r'C:/Users/jacki/Downloads/AU26_Lab2_Data_Ramps.mat')
square_data = ntk.LabChartDataset(r'C:/Users/jacki/Downloads/AU26_Lab2_Data_Pulses.mat')
```

```python
#style
mpl.rcParams['font.family'] = 'arial'
mpl.rcParams['font.size'] = 14
mpl.rcParams['axes.spines.top']= False
mpl.rcParams['axes.spines.right']= False
mpl.rcParams['axes.spines.left']= False
mpl.rcParams['axes.spines.bottom']= False
```

# raw data plots

```python
# get data for 150 mv and 1.5 V
# each should have many rows, row 0 being the light and 2 being the smoothed electrode recording from the fly eye. sampling rate was 2k samples/second
square150mV = square_data.get_block(16) 
square1500mV = square_data.get_block(138) 
```

```python
#150 mV raw data plot
fig, ax = plt.subplots(nrows = 2,ncols = 1, sharex=True, figsize = (3,4))
ax[0].plot(square150mV[0], color = "#E97132", linewidth=2)
ax[0].set_ylim(-0.2, 2)

ax[1].plot(square150mV[2]-0.15, color = "#0F9ED5", linewidth=2)
ax[1].set_ylim(-1, 0.5)

xticks=np.array([0,1,2])
plt.xticks(xticks*2000,xticks)
plt.xlim(0, 4000)

plt.show()
```

```python
#1.5V raw data plot
fig, ax = plt.subplots(nrows = 2,ncols = 1, sharex=True, figsize = (3,4))
ax[0].plot(square1500mV[0], color = "#E97132", linewidth=2)
ax[0].set_ylim(-0.2, 2)

ax[1].plot(square1500mV[2]-0.5, color = "#0F9ED5", linewidth=2)
ax[1].set_ylim(-1, 0.5)

xticks=np.array([0,1,2])
plt.xticks(xticks*2000,xticks)
plt.xlim(0, 4000)

plt.show()
```

```python
# get data for ramp up 2V/s and 8V/s
# each should have many rows, row 0 being the light and 2 being the electrode recording from the fly eye. sampling rate was 2k samples/second
ramp2 = ramp_data.get_block(24) 
ramp8 = ramp_data.get_block(52) 
```
